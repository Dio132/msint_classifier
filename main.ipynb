```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from sklearn.utils import shuffle
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
from tensorflow.keras import optimizers
from tensorflow.keras.layers import Dense
from tensorflow.keras.datasets import mnist
```

```python
np.random.seed(1212)
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train = x_train/255
x_test = x_test/255
```

```python
x_train_flat = x_train.reshape(x_train.shape[0], (28*28))
x_test_flat = x_test.reshape(x_test.shape[0], (28*28))
```

```python
n_input = 784  # Number of features (28x28 pixels flattened)
n_hidden_1 = 300
n_hidden_2 = 100
n_hidden_3 = 100
n_hidden_4 = 200
num_digits = 10  # Output classes (digits 0–9)
```

```python
# Build model using Sequential API
model = Sequential([
    Dense(n_hidden_1, activation='relu', input_shape=(n_input,), name="Hidden_Layer_1"),
    Dense(n_hidden_2, activation='relu', name="Hidden_Layer_2"),
    Dense(n_hidden_3, activation='relu', name="Hidden_Layer_3"),
    Dense(n_hidden_4, activation='relu', name="Hidden_Layer_4"),
    Dense(num_digits, activation='softmax', name="Output_Layer")
])
```

```python
# Print model summary
model.summary()
```

```python
# Hyperparameters
learning_rate = 0.1
training_epochs = 5
batch_size = 100

# Optimizer
sgd = optimizers.SGD(learning_rate=learning_rate)
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='sgd',
              metrics=['accuracy'])
```

```python
# Fit the model with your data
history1 = model.fit(x_train_flat, y_train,
                     batch_size = batch_size,
                     epochs = training_epochs,
                     verbose = 2,
                     )
```

```python
test_loss, test_accuracy = model.evaluate(x_test_flat, y_test, verbose=2)
```

```python
model.save("mnist_model.h1")
```
